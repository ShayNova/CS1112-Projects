## Problem 1 Approximating the cube root

```python
# user must enter a value
v = float(input('Enter a positive value: '))
print(f'Estimate the cube root of {v:.1f}') 
s3 = 1.0
s2 = 1.0
step = 0

print('-------------------------------------------')
print('Step No.   Average over 3   Average over 2')
print('-------------------------------------------')
# step 0
print(f'{step: 5.0f} {s3: 15.4f} {s2: 15.4f}')

# step 1
s_new1 = 1/3*(s3+s3+(v/s3**2))
s_new2 = 1/2*(s2+(v/s2**2))
print(f'{step+1: 5.0f} {s_new1: 15.4f} {s_new2: 15.4f}')
s3 = s_new1
s2 = s_new2

# step 2
s_new1 = 1/3*(s3+s3+(v/s3**2))   
s_new2 = 1/2*(s2+(v/s2**2))
print(f'{step+2: 5.0f} {s_new1: 15.4f} {s_new2: 15.4f}')
s3 = s_new1
s2 = s_new2

# step 3
s_new1 = 1/3*(s3+s3+(v/s3**2))
s_new2 = 1/2*(s2+(v/s2**2))
print(f'{step+3: 5.0f} {s_new1: 15.4f} {s_new2: 15.4f}')
s3 = s_new1
s2 = s_new2

# step 4
s_new1 = 1/3*(s3+s3+(v/s3**2))
s_new2 = 1/2*(s2+(v/s2**2))
print(f'{step+4: 5.0f} {s_new1: 15.4f} {s_new2: 15.4f}')
s3 = s_new1
s2 = s_new2

# step 5
s_new1 = 1/3*(s3+s3+(v/s3**2))
s_new2 = 1/2*(s2+(v/s2**2))
print(f'{step+5: 5.0f} {s_new1: 15.4f} {s_new2: 15.4f}')
```

1. Which averaging method do you prefer?  Why?

I prefer the average over 2 because it's more accurate.

2. How well does this approximation method work with increasing $v$?

As v increases, the approximation gradually becomes less accurate because there are few trials.

## Problem 2 Book Seller
This problem is modified from our textbook _Think Python_ [Section 2.10, Exercise 2 Problem 2](https://greenteapress.com/thinkpython2/html/thinkpython2003.html#sec25)

```python
# user must enter price
price = float(input('Enter the cover price: '))
# user must enter number of copies
copies = int(input('Enter number of copies: '))
# user must enter if they're a bookstore or not, and I turn the string into lowercase
customer = str(input('Is this a bookstore (y/n): ').lower())

# if user is bookstore, they must enter discount
if customer == 'y':
    discount = float(input('Enter the bookstore discount percentage: '))
# if user is not a bookstore, they must enter tax rate percentage
else:
    tax = float(input('Enter tax rate percentage: '))

# print the bill for the customer
print(' ')
print('-----Bookstore Bill-----')

# book cost
if customer == 'y':
    discount1 = price - (price*(discount/100.))
    print(f'Book cost: {copies*discount1:13.2f}')
else:
    print(f'Book cost: {copies*price:13.2f}')
    
# shipping cost
shipping = 5.
if copies == 1:
    print(f'Shipping cost: {shipping:9.2f}')
else:
    shipping1 = shipping+ 0.95*(copies-1)
    print(f'Shipping cost: {shipping1:9.2f}')
    
# taxes
if customer == 'y':
    print('Tax:                0.00')    
elif customer != 'y' and copies == 1:
    tax1 = (price+shipping)*(tax/100)
    print(f'Tax: {tax1:19.2f}')
else:
    tax2 = ((price*copies)+shipping1)*(tax/100)
    print(f'Tax: {tax2:19.2f}')
    
# total cost
if customer == 'y' and copies == 1:
    print(f'Total cost: {discount1+shipping:12.2f}')
elif customer == 'y' and copies != 1:
    print(f'Total cost: {copies*discount1+shipping1:12.2f}')
elif customer != 'y' and copies == 1:
    print(f'Total cost: {price+shipping+tax1:12.2f}')
else:
    print(f'Total cost: {copies*price+shipping1+tax2:12.2f}')
```
