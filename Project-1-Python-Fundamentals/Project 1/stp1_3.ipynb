```python
import random
import matplotlib.pyplot as plt

"""
  This program positions a light source and a box in a room
  and determines the extent of the shadow cast by the box.
"""

# Set room size
# Lower-left corner at (0,0), upper-right corner at (xmax,ymax)
# xmax, ymax must be >= 10
xmax= 10
ymax= 12

# Set up figure window
plt.close('all')    # Close all previous figure windows
plt.figure()        # Start a new figure window
plt.gca().set_aspect('equal', adjustable='box')  # Ensure equal aspect ratio
plt.axis([0, xmax, 0, ymax])  # x-axis limits [0,xmax], y-axis limits [0,ymax]
plt.box(True)

# Top left corner of box, point T
xt= 5
yt= 7
plt.plot(xt, yt, 'bo')             # Format: Blue circle
plt.plot([xt, xt], [1, yt], 'k:')  # Format: blacK dotted line
plt.text(xt, yt, '  T')

### Do not change the code above ###
###   other than the room size   ###
####################################
### Modify the code below ##########

# Light source, point L
xL= xt * random.random()      # the range limit for both axes
yL= yt + 5 * random.random()
plt.plot(xL, yL, 'r*')   # Format: Red asterisk
plt.text(xL, yL, '  L')

# Draw a cyan solid line from point T to the point (xmax,yL)
plt.plot([xt, xmax], [yt, yt], 'c-')   # Format: Cyan line

# User-clicked point
plt.title('Click on the dotted line below')
plt.show()  # show the plot
point= plt.ginput(1)  # Accepts one user click; note time-out of 30 seconds
xu, yu= point[0]      # Stores x-coord of click in variable xu,
                      #        y-coord of click in variable yu
plt.plot(xu, yu, 'm+')  # Format: Magenta cross
plt.text(xt, yu, '  U') # Changed xu to xt to keep bottom edge accurate
message_to_show = f'You clicked at ({xu:.1f},{yu:.1f})'
plt.title(message_to_show)

# Draw the rest of the box
plt.plot([xmax, xmax], [yt, yu], 'c-')
plt.plot([xmax, xt], [yu, yu], 'c-')
plt.plot([xt, xt], [yu, yt], 'c-')

# Calculate the slope
plt.plot([xL, xt], [yL, yu], 'r--')
m = (yu-yL)/(xt-xL)   # slope formula
yR = m*(xmax-xL)+yL   # right wall y-coordinate
xB = xL-(yL/m)        # bottom wall x-coordinate

# Draw length of light line
plt.plot([xL, xB], [yL, 0], 'r--')

# Draw the shadow line
if yR > 0:
    plt.plot([xmax, xmax], [yu, yR], 'b-', linewidth=4)
else:
    plt.plot([xmax, xmax], [yu, 0], 'b-', linewidth=4)
    plt.plot([xmax, xB], [0, 0], 'b-', linewidth=4)

# Light coordinate title
new_message = f'Light coordinates: ({xL:.1f},{yL:.1f})'
plt.title(new_message)

####################################
### Do not change the code below ###

plt.show()
```
